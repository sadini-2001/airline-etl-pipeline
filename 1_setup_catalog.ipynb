**Create Airline Catalog**

```python
%sql
CREATE CATALOG IF NOT EXISTS airline;
```

```python
%sql
USE CATALOG airline;
```

**Create Bronze, Silver, Gold Schemas**

```python
%sql
CREATE SCHEMA IF NOT EXISTS airline.bronze;
CREATE SCHEMA IF NOT EXISTS airline.silver;
CREATE SCHEMA IF NOT EXISTS airline.gold;
```

**Verify all schemas created**

```python
%sql
SHOW DATABASES FROM airline;
```
