```python
import pandas as pd
import os

# ── TWO PATHS ──────────────────────────────────────
INPUT_FOLDER  = r"C:\Users\sadin\OneDrive\Desktop\Sadini\projects\ETL-DATABRICKS\AirPlane Dataset"   # folder where my 4 CSVs are
OUTPUT_FOLDER = r"C:\Users\sadin\OneDrive\Desktop\Sadini\projects\ETL-DATABRICKS\AirPlane Dataset\preprocessed"  # where daily files will be saved
# ───────────────────────────────────────────────────────────────────

COLS_TO_KEEP = [
    "FlightDate", "Reporting_Airline", "Tail_Number",
    "Flight_Number_Reporting_Airline",
    "Origin", "Dest", "OriginCityName", "DestCityName",
    "OriginState", "DestState",
    "CRSDepTime", "DepTime", "DepDelay",
    "CRSArrTime", "ArrTime", "ArrDelay",
    "ActualElapsedTime",
    "Cancelled", "CancellationCode", "Diverted",
    "CarrierDelay", "WeatherDelay", "NASDelay",
    "SecurityDelay", "LateAircraftDelay"
]

files = [
    "On_Time_Reporting_Carrier_On_Time_Performance_(1987_present)_2024_1.csv",
    "On_Time_Reporting_Carrier_On_Time_Performance_(1987_present)_2024_2.csv",
    "On_Time_Reporting_Carrier_On_Time_Performance_(1987_present)_2024_3.csv",
    "On_Time_Reporting_Carrier_On_Time_Performance_(1987_present)_2024_4.csv",
]

os.makedirs(OUTPUT_FOLDER, exist_ok=True)
total_files = 0

for f in files:
    full_path = os.path.join(INPUT_FOLDER, f)   # combines folder + filename
    print(f"\nReading {full_path}...")
    df = pd.read_csv(full_path, usecols=COLS_TO_KEEP, low_memory=False)
    print(f"  {len(df):,} rows · {df.memory_usage(deep=True).sum() / 1e6:.1f} MB in memory")

    for date, group in df.groupby("FlightDate"):
        folder = os.path.join(OUTPUT_FOLDER, str(date))
        os.makedirs(folder, exist_ok=True)
        group.to_csv(os.path.join(folder, f"flights_{date}.csv"), index=False)
        total_files += 1

    print(f"  {df['FlightDate'].nunique()} daily files created")
    del df

print(f"\nDone. Total daily files: {total_files}")
print(f"Output folder: {OUTPUT_FOLDER}")
```
